```python
import model as md
import utils
import os
import torch

utils.organize_test_folder(os.path.join('..', 'data', 'asl_alphabet_test', 'asl_alphabet_test'))

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('using ', device, 'device')
print(torch.__version__)
```

```python
model = md.CNN(device = device)

train_root = os.path.join('..', 'data', 'asl_alphabet_train', 'asl_alphabet_train')
train_loader, val_loader = utils.load_train_data(train_root)
```

```python
model.my_train(train_loader=train_loader, val_loader=val_loader, epochs=10)
```

```python
model_path = os.path.join('..', 'models', 'cnn_asl.pth')

# 3. Сохраняем состояние модели (только веса)
torch.save(model.state_dict(), model_path)
print(f"Модель успешно сохранена в: {model_path}")
```
