```python
import model as md
import utils
import os
import torch

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('using ', device, 'device')
print(torch.__version__)
```

```python
model = md.CNN(device = device)

train_root = os.path.join('..', 'data', 'Train_Alphabet')
test_root = os.path.join('..', 'data', 'Test_Alphabet')

train_loader, val_loader = utils.load_train_data(train_root)
test_loader = utils.load_test_data(test_root)
```

```python
model.my_train(train_loader=train_loader, val_loader=val_loader, epochs=20)
```

```python
model.test(test_loader)
```

```python
model_path = os.path.join('..', 'models', 'cnn_asl.pth')

# 3. Сохраняем состояние модели (только веса)
torch.save(model.state_dict(), model_path)
print(f"Модель успешно сохранена в: {model_path}")
```
